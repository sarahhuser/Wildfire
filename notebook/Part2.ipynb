```python
import pandas as pd

# 1. Loading data

fire = pd.read_csv("../data/fire_clean.csv", sep = ",") #../ ist nötig, damit man aus dem notbook ordner kommmt und dann in den datd ordner

print("Data loaded successfully")

display(fire.head(5))
```

```python
# 2. Histogram

import xarray as xr
import matplotlib.pyplot as plt

# Load the fire dataset

ds = xr.fire.open_dataset("size_ha")

ds.fire.plot.hist(bins=50,  histtype="stepfilled", alpha=0.8)
plt.xlabel("Size (ha)")
plt.ylabel("fire_id")
plt.title("Distribution of fire size")

plt.show()
```
